**BFS**

```python
from collections import deque
import networkx as nx
import matplotlib.pyplot as plt
# Definee the Graph
graph = {
    'A': ['B','C'],
    'B': ['D','E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

#BFS Function
def bfs(graph,start):
    visited =set()
    queue = deque([start])
    traversal = []
    visited.add(start)
    while queue:
       node = queue.popleft()
       traversal.append(node)
       for neighbor in graph [node]:
           if neighbor not in visited:
               visited.add(neighbor)
               queue.append(neighbor)

    return traversal

#Run BFS

result = bfs(graph,'A')
print("BFS Traversal:")
print("->".join(result))

#Graphical Representation
#Create a directed graph

G = nx.DiGraph()
#Add edges from the dictionary
for node in graph:
  for neighbor in graph[node]:
    G.add_edge(node,neighbor)
#Define positions for better tree layout
pos ={
    'A': (0,3),
    'B': (-1,2),
    'C': (1,2),
    'D': (-1.5,1),
    'E': (-0.5,1),
    'F': (1,1),
    'G': (-0.5,0)
}
#Draw the Graph
plt.figure(figsize=(8,6))
nx.draw(
    G,pos,
    with_labels=True,
    node_color='skyblue',
    node_size = 2500,
    font_size = 14,
    edge_color='black',
    arrows = True,
    width = 2
)
plt.title("BFS Graph")
plt.show()
```

**DFS**

```python
from collections import deque
import networkx as nx
import matplotlib.pyplot as plt

# Define the Graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

# DFS Function (Recursive)
def dfs(graph, node, visited=None, traversal=None):
    if visited is None:
        visited = set()

    if traversal is None:
        traversal = []

    visited.add(node)
    traversal.append(node)

    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited, traversal)

    return traversal

# Run DFS
result = dfs(graph, 'A')

print("DFS Traversal:")
print(" -> ".join(result))

# Graphical Representation
G = nx.DiGraph()

# Add edges from the dictionary
for node in graph:
    for neighbor in graph[node]:
        G.add_edge(node, neighbor)

# Define positions for tree layout
pos = {
    'A': (0, 3),
    'B': (-1, 2),
    'C': (1, 2),
    'D': (-1.5, 1),
    'E': (-0.5, 1),
    'F': (1, 1),
    'G': (-0.5, 0)
}

# Draw the Graph
plt.figure(figsize=(8, 6))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_color='purple',
    node_size=2500,
    font_size=14,
    font_color='white',
    edge_color='black',
    arrows=True,
    width=2
)

plt.title("DFS Graphical Representation", fontsize=18)
plt.axis('off')
plt.show()
```
